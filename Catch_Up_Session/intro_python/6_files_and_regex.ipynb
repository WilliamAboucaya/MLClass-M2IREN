## 1. Reading files

Python opens files with the built-in `open()` function. The recommended pattern uses `with`, which automatically closes the file for you once you're done with it, even if something goes wrong inside the block:

```python
with open("example.txt", "w") as f:
    f.write("Hello, file!\nThis is a second line.\nAnd a third one.")
```

(We just used `open()` in **write mode** (`"w"`) to create a small text file to play with. More on writing in the next section.)

Now let's read it back:

```python
with open("example.txt", "r") as f:   # "r" = read mode (also the default if omitted)
    content = f.read()

print(content)
print(type(content))
```

`.read()` returns the entire file as a single string. If you'd rather work line by line, `.readlines()` gives you a list of lines, or you can loop directly over the file object:

```python
with open("example.txt", "r") as f:
    lines = f.readlines()

print(lines)
```

```python
with open("example.txt", "r") as f:
    for line in f:
        print(line.strip())   # .strip() removes the trailing newline character
```

Looping directly over the file (like the last example) is the most common pattern for large files, since it reads one line at a time instead of loading everything into memory at once.

## 2. Writing files

You already saw write mode (`"w"`) above which creates the file if it doesn't exist, and **overwrites** it completely if it does. If you instead want to add to an existing file without erasing it, use append mode (`"a"`):

```python
with open("example.txt", "a") as f:
    f.write("\nA fourth line, added later.")

with open("example.txt", "r") as f:
    print(f.read())
```

Writing line by line from a list is a common pattern. Note that `.write()` doesn't add newlines automatically, so you need to include them yourself:

```python
fruits = ["apple", "banana", "cherry"]

with open("fruits.txt", "w") as f:
    for fruit in fruits:
        f.write(fruit + "\n")

with open("fruits.txt", "r") as f:
    print(f.read())
```

## 3. Regex basics

**Regex** (short for "regular expression") is a mini-language for describing patterns in text. It is extremely useful for tasks like "does this look like an email address?" or "extract every number from this string". Python's regex tools live in the built-in `re` module.

```python
import re

text = "My phone number is 0612345678."

match = re.search(r"\d+", text)   # \d+ means "one or more digits"
print(match)
print(match.group())   # the actual matched text
```

A few building blocks you'll use most often:

| Pattern | Meaning                                           |
|---------|---------------------------------------------------|
| `\d`    | a digit (0-9)                                     |
| `\w`    | a "word" character (letter, digit, or underscore) |
| `\s`    | a whitespace character                            |
| `+`     | one or more of the previous thing                 |
| `*`     | zero or more of the previous thing                |
| `.`     | any character                                     |

```python
text = "Contact: alice@example.com or bob99@test.org"

emails = re.findall(r"\w+@\w+\.\w+", text)   # find ALL matches, as a list
print(emails)
```

The `r` before the string (e.g. `r"\d+"`) marks it as a **raw string**, which prevents Python from interpreting backslashes specially. It's a habit worth keeping any time you write a regex pattern.

## 4. Replacing text with `re.sub`

`re.sub(pattern, replacement, text)` finds every match of a pattern and replaces it, very handy for cleaning up messy text data.

```python
text = "Call me at 06-12-34-56-78 or 06 98 76 54 32."

# remove all dashes and spaces from phone numbers
cleaned = re.sub(r"[-\s]", "", text)
print(cleaned)
```

`[-\s]` here means "a dash or a whitespace character". Square brackets define a **character class**, i.e. "match any one of these characters."

You can also use `re.sub` to normalize text, e.g. collapsing multiple spaces into one:

```python
messy = "This   text   has    irregular   spacing."
cleaned = re.sub(r"\s+", " ", messy)
print(cleaned)
```
