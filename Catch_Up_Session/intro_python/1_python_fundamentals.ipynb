## 1. Variables and assignment
 
A **variable** is a name that stores a value so you can reuse it later. You create one with `=`, the assignment operator.

*N.B.: Here, `=` should not be read as "equals" in the math sense, rather "put this value into that box".*

```python
age = 29
city = "Paris"
print(age)
print(city)
```

A few naming rules:
- Names can contain letters, digits, and underscores, but can't start with a digit (`age2` is fine, `2age` is not)
- Names are case-sensitive (`Age` and `age` are different variables)
- By convention, Python variable names are `lower_case_with_underscores` (snake case)
- Avoid using names of built-in functions (e.g. don't name a variable `print`)

Variables can be reassigned at any time:

```python
age = 29
print(age)
age = 30
print(age)  # age has changed its value
```

**Important**: If you want to signal that an element of your code should not be executed (for example, a comment explaining what is happening), the most common way to signal it to the Python interpreter is to prepend your comment wit a `#`

For example: 

```python
1 + 2 + 3  # + 4
```

## 2. Core data types
 
Every value in Python has a **type**. The four you'll use constantly are:
 
| Type | Meaning | Example |
|---|---|---|
| `int` | whole number | `1234` |
| `float` | decimal number | `3.14` |
| `str` | text (string) | `"hello"` |
| `bool` | true/false | `True`, `False` |
 
There's also a special value, `None`, which represents "nothing" or "no value yet".


```python
age = 24            # int
height = 1.68       # float
name = "Chandini"   # str
is_student = True   # bool
job = None          # "no value"

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
print(type(job))
```

`type()` is another built-in function: it tells you the type of whatever you give it. You'll use it a lot when debugging.
 
A couple of things worth noticing about strings:
- They can be written with single `'...'` or double `"..."` quotes, both work the same
- You can check their length with `len(...)`

```python
message = "Hello Python"
print(len(message))
```

## 3. Basic operators
 
### Arithmetic operators

```python
a = 7
b = 2
 
print(a + b)   # addition       -> 9
print(a - b)   # subtraction    -> 5
print(a * b)   # multiplication -> 14
print(a / b)   # division       -> 3.5 (always returns a float)
print(a // b)  # floor division -> 3 (rounds down to an int)
print(a % b)   # modulo (remainder) -> 1
print(a ** b)  # power          -> 49
```

### Comparison operators
 
These always return a `bool` (`True` or `False`):

```python
print(5 == 5)   # equal to
print(5 != 3)   # not equal to
print(5 > 3)    # greater than
print(5 < 3)    # less than
print(5 >= 5)   # greater than or equal to
print(5 <= 4)   # less than or equal to
```

**Common beginner mistake**: `=` assigns a value, `==` compares two values. Mixing them up is one of the most frequent bugs in early Python code.

### Logical operators
 
Used to combine booleans:

```python
is_sunny = True
is_weekend = False
 
print(is_sunny and is_weekend)  # True only if both are True
print(is_sunny or is_weekend)   # True if at least one is True
print(not is_sunny)             # flips True <-> False
```

### A word on strings and f-strings
 
Strings can be combined ("concatenated") with `+`, but the cleanest way to mix text and variables is an **f-string**:

```python
name = "Alex"
age = 25
 
# concatenation (works, but clunky with numbers)
print("My name is " + name + " and I am " + str(age) + " years old.")
 
# f-string (cleaner, and handles types automatically)
print(f"My name is {name} and I am {age} years old.")
```

You'll see f-strings everywhere in real Python code, they're the standard way to build strings with variables inside.

## 4. Type conversion (casting)
 
Sometimes you need to convert a value from one type to another. This is called **casting**, and it's very common when working with user input or data read from files (more on that in notebook 6).

```python
x = "42"
print(type(x))
 
y = int(x)
print(type(y))
print(y + 8)   # now this works, since y is a number
```

The main conversion functions:

```python
print(int("10"))       # str -> int
print(float("3.14"))   # str -> float
print(str(99))         # int -> str
print(bool(0))         # -> False
print(bool(1))         # -> True
print(bool(""))        # -> False (empty string)
print(bool("hello"))   # -> True (non-empty string)
```

Careful: `int("3.14")` will raise an error: you can't convert a decimal-looking string directly to `int`. You'd need `int(float("3.14"))` if you really want to truncate it.

## 5. Conditionals: if / elif / else
 
Conditionals let your program make decisions and run different code depending on a condition.

```python
temperature = 37
 
if temperature > 25:
    print("It's hot!")
```

The line inside the `if` block is **indented** (usually 4 spaces). This isn't just style: in Python, indentation is how you define blocks of code. Get it wrong and your code either won't run, or won't do what you expect.
 
You can chain conditions with `elif` ("else if") and finish with `else`:

```python
temperature = 37
 
if temperature > 25:
    print("It's hot!")
elif temperature > 15:
    print("It's mild.")
else:
    print("It's cold!")
```

Python checks conditions top to bottom and runs the **first** block that matches, the rest are skipped even if they would also be true.
 
You can also combine conditions using the logical operators from earlier:

```python
age = 20
has_ticket = True
 
if age >= 18 and has_ticket:
    print("You can enter.")
else:
    print("You cannot enter.")
```
