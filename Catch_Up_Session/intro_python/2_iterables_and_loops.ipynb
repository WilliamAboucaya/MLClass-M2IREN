## 0. Recap & motivation

In notebook 1, every variable held a **single** value: one number, one string, one boolean. In practice, you'll almost always work with **collections** of values: a list of customer names, a table of prices, a set of unique IDs.

## 1. Lists

A **list** is an ordered, changeable collection of values, written with square brackets `[]`.

````python
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(type(fruits))
print(len(fruits))
````

Lists can mix types, though in practice you'll mostly use lists of a single type:

````python
mixed = ["William", 29, True]
print(mixed)
````

### Indexing and slicing

Each element has a position, or **index**, starting at `0`.

````python
fruits = ["apple", "banana", "cherry", "date"]
 
print(fruits[0])    # first element -> "apple"
print(fruits[2])    # third element -> "cherry"
print(fruits[-1])   # last element -> "date"
print(fruits[1:3])  # slice: elements from index 1 up to (not including) 3
````

### Modifying a list

````python
fruits = ["apple", "banana", "cherry"]
 
fruits.append("date")       # add at the end
print(fruits)
 
fruits.remove("banana")     # remove by value
print(fruits)
 
fruits[0] = "avocado"       # replace by index
print(fruits)
````

## 2. Tuples

A **tuple** looks like a list but uses parentheses `()`, and it's **immutable**: once created, you can't change, add, or remove elements.

````python
coordinates = (48.85, 2.35)
print(coordinates)
print(coordinates[0])
````

````python
coordinates = (48.85, 2.35)
coordinates[0] = 50   # this will raise an error
````

Why use a tuple instead of a list, then? Mainly when you want to signal "this shouldn't change" (e.g. fixed coordinates, RGB color values, or a pair of values returned together from a function).

## 3. Dictionaries

A **dictionary** stores **key -> value** pairs, written with curly braces `{}`. Instead of accessing elements by position (like a list), you access them by **key**.

````python
teacher = {
    "name": "William",
    "age": 29,
    "is_here": True
}

print(teacher)
print(teacher["name"])
print(teacher["age"])
````

Adding or updating a key works just like assigning a variable:

````python
teacher = {"name": "William", "age": 29}

teacher["age"] = 30          # update existing key
teacher["city"] = "Paris"    # add a new key

print(teacher)
````

Useful methods:

````python
teacher = {"name": "William", "age": 29, "city": "Paris"}

print(teacher.keys())     # all keys
print(teacher.values())   # all values
print(teacher.items())    # key-value pairs
print("name" in teacher)  # check if a key exists -> True
````

Dictionaries are everywhere in data work. Think of a single row of data described by column names, or a JSON file loaded from an API.

## 4. Sets

A **set** is an **unordered** collection of **unique** values, written with curly braces too (but no key-value pairs, just values).

````python
unique_ids = {101, 102, 103, 102, 101}
print(unique_ids)   # duplicates are automatically removed
````

Sets are handy for two things in particular: removing duplicates, and checking membership very quickly

````python
colors_seen_in_photo_1 = {"red", "green", "blue"}
colors_seen_in_photo_2 = {"blue", "yellow"}

print(colors_seen_in_photo_1 & colors_seen_in_photo_2)  # intersection
print(colors_seen_in_photo_1 | colors_seen_in_photo_2)  # union
print(colors_seen_in_photo_1 - colors_seen_in_photo_2)  # difference
````

## 5. `for` loops

A `for` loop repeats a block of code once **for each element** of a collection.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

You can loop over a range of numbers with `range()`:

````python
for i in range(5):       # 0, 1, 2, 3, 4
    print(i)
````

````python
for i in range(2, 10, 2):  # start, stop, step -> 2, 4, 6, 8
    print(i)
````

Looping over a dictionary usually means looping over its `.items()`:

````python
teacher = {"name": "William", "age": 29, "city": "Paris"}
 
for key, value in teacher.items():
    print(key, "->", value)
````

Combining a loop with a conditional is extremely common:

````python
numbers = [3, 8, 15, 22, 4, 11]
 
for number in numbers:
    if number % 2 == 0:
        print(number, "is even")
    else:
        print(number, "is odd")
````

## 6. `while` loops
 
A `while` loop repeats **as long as a condition stays true**. It is mainly useful when you don't know in advance how many times you'll need to loop.

````python
count = 0
 
while count < 5:
    print(count)
    count = count + 1   # this line is crucial! without it, the loop never ends
````

**Important**: Forgetting to update the condition variable is the classic beginner mistake here: it creates an **infinite loop**. If that happens in Jupyter, you can interrupt/stop the cell from the toolbar.

You can also exit a loop early with `break`, or skip an iteration with `continue`:

````python
count = 0

while True:                # loops forever, unless we break out
    if count == 3:
        break
    print(count)
    count = count + 1
````

## 7. List comprehensions

A **list comprehension** is a compact way to build a list from a loop, in a single line. It's not new functionality, it's a more "Pythonic" way to write patterns you already know.

Compare the two equivalent approaches:

````python
# classic way
squares = []
for x in range(6):
    squares.append(x ** 2)
print(squares)
````

````python
# list comprehension
squares = [x ** 2 for x in range(6)]
print(squares)
````

You can add a condition too:

````python
numbers = [3, 8, 15, 22, 4, 11]

even_numbers = [n for n in numbers if n % 2 == 0]
print(even_numbers)
````

You'll see this style constantly in real-world Python (and later, in similar form, with pandas). It's worth getting comfortable reading it, even if you prefer writing classic loops at first (both are perfectly valid).
