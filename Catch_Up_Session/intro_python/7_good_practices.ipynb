## 1. Readability and naming conventions

Python has an official style guide called **PEP 8**. You don't need to memorize it, but a few conventions are worth adopting from day one, because they're what virtually all Python code (and all the libraries you'll use) follows.

### Naming conventions

| What | Convention | Example |
|---|---|---|
| Variables & functions | `snake_case` (lowercase, underscores) | `student_age`, `compute_total()` |
| Constants | `UPPER_SNAKE_CASE` | `MAX_RETRIES = 3` |
| Classes (you'll meet these later) | `PascalCase` | `StudentRecord` |
| "Private"-ish names (by convention only) | leading underscore | `_internal_helper()` |

````python
# Good: clear, conventional names
student_age = 25
MAX_ATTEMPTS = 5

def compute_average(values):
    return sum(values) / len(values)


# Avoid: unclear or non-conventional names
a = 25                      # what is "a"?
maxAttempts = 5              # camelCase isn't the Python convention
def computeAvg(v):           # same issue, plus a cryptic parameter name
    return sum(v) / len(v)
````

### Naming things well (not just consistently)

Beyond the *format* of names, their *content* matters just as much:
- Prefer `student_age` over `age` if there's any ambiguity about whose age it is.
- Prefer full words over cryptic abbreviations: `temperature` over `temp` (unless the meaning is truly obvious in context), `dataframe` or `df` over `d`.
- Boolean variables read better with a `is_`/`has_`/`should_` prefix: `is_valid`, `has_permission`, rather than just `valid` or `permission`.
- Function names should usually be verbs (they *do* something): `calculate_total()`, `filter_students()` as opposed to variable names, which are usually nouns.

````python
# Good: names communicate intent
is_eligible = student_age >= 18
has_discount = True

def filter_adults(people):
    return [p for p in people if p["age"] >= 18]


# Avoid: technically correct, but unclear
flag = student_age >= 18
x = True

def f(p):
    return [i for i in p if i["age"] >= 18]
````

### Other quick PEP 8 habits

- Use 4 spaces for indentation (not tabs). Jupyter does this by default.
- Add spaces around operators: `x = a + b`, not `x=a+b`.
- Keep lines reasonably short (PEP 8 suggests ~79 characters, many teams use up to ~100. The exact number matters less than staying consistent).
- Leave a blank line or two between logical sections of code, similar to paragraphs in writing.

## 2. Comments and docstrings

**Comments** (starting with `#`) explain *why* something is done, not *what* the code does line by line: the code itself should already show the "what," if it's well-named.

```python
# Bad comment: just restates the code
x = x + 1  # add 1 to x

# Good comment: explains a non-obvious reason
retry_count = retry_count + 1  # server sometimes drops the first request, so we retry
```

A **docstring** is a special string placed right after a function's `def` line, describing what it does, its parameters, and what it returns. It's accessed with `help()` or `function.__doc__`, and is what most editors show as a tooltip.

````python
def compute_bmi(weight_kg, height_m):
    """
    Compute the Body Mass Index (BMI) for a person.

    Parameters:
        weight_kg (float): weight in kilograms.
        height_m (float): height in meters.

    Returns:
        float: the computed BMI.
    """
    return weight_kg / (height_m ** 2)


help(compute_bmi)
````

A short, single-line docstring is often enough for simple functions, you don't need the full parameter/return breakdown every time:

````python
def square(x):
    """Return the square of x."""
    return x ** 2
````

As a rule of thumb: every function you write for others to use (or that you'll reuse yourself later) should have at least a one-line docstring. Comments should be used sparingly, reserved for the genuinely non-obvious parts.

## 3. Type hints

**Type hints** let you annotate a function's parameters and return value with the types they're expected to be. Python doesn't enforce them at runtime (they're documentation, *not a guarantee*) but they make a function's contract obvious at a glance, and editors/tools use them to catch mistakes before you even run the code.

````python
def compute_average(values: list[float]) -> float:
    return sum(values) / len(values)


print(compute_average([2.0, 4.0, 6.0]))
````

The syntax is `parameter: type` for inputs, and `-> type` for the return value. A few common annotations:

````python
def greet(name: str, times: int = 1) -> str:
    return (f"Hello, {name}! " * times).strip()


def is_adult(age: int) -> bool:
    return age >= 18


def get_students(names: list[str]) -> dict[str, int]:
    return {name: len(name) for name in names}
````

Type hints and docstrings complement each other rather than replace one another: the hint tells you *what type* is expected, the docstring tells you *what the function does* and *why*.

````python
def apply_discount(price: float, discount: float) -> float:
    """Return the price after applying a discount (as a fraction, e.g. 0.1 for 10%)."""
    return price * (1 - discount)
````

Remember these are hints, not enforced rules: Python will still run `compute_average(["a", "b"])` without complaining, and only fail once it tries to actually `sum()` the strings. The benefit is mostly for humans (and editor tooling), not the interpreter itself.

## 4. Avoiding code duplication (DRY)

**DRY** stands for "Don't Repeat Yourself." If you find yourself copy-pasting the same few lines of code with only small tweaks, that's a strong signal it should become a function instead. Duplicated code is a maintenance trap: fix a bug in one copy, and it's easy to forget the other copies exist.

````python
# Repetitive: the same calculation copy-pasted three times
price_1 = 100
discount_1 = 0.1
final_price_1 = price_1 * (1 - discount_1)

price_2 = 50
discount_2 = 0.2
final_price_2 = price_2 * (1 - discount_2)

price_3 = 200
discount_3 = 0.15
final_price_3 = price_3 * (1 - discount_3)

print(final_price_1, final_price_2, final_price_3)
````

````python
# DRY: the logic lives in one place
def apply_discount(price: float, discount: float) -> float:
    """Return the price after applying a discount (as a fraction)."""
    return price * (1 - discount)


final_price_1 = apply_discount(100, 0.1)
final_price_2 = apply_discount(50, 0.2)
final_price_3 = apply_discount(200, 0.15)

print(final_price_1, final_price_2, final_price_3)
````

If a bug is later found in the discount formula, the DRY version needs a fix in exactly one place. The repetitive version needs the same fix applied (correctly!) in every copy -> easy to miss one.

The same idea applies to loops: if you're about to write a near-identical loop body for several different lists, ask whether a function called from each loop (or a single loop over a combined structure) would remove the duplication.

## 5. One function, one task

A good function does **one thing**, and its name should make that one thing obvious. When a function starts doing several unrelated things at once, it becomes harder to name well, harder to test, and harder to reuse, as you can't reuse "half" of a function.

````python
# Doing too much at once: loading, cleaning, AND computing, all in one function
def process_scores(raw_scores: list[str]) -> float:
    # cleaning: convert strings to numbers, ignore invalid entries
    cleaned = []
    for s in raw_scores:
        try:
            cleaned.append(float(s))
        except ValueError:
            pass
    # computing: return the average
    return sum(cleaned) / len(cleaned)


print(process_scores(["80", "92", "not_a_number", "75"]))
````

````python
# One task per function: easier to name, test, and reuse independently
def clean_scores(raw_scores: list[str]) -> list[float]:
    """Convert a list of strings to floats, dropping any that aren't valid numbers."""
    cleaned = []
    for s in raw_scores:
        try:
            cleaned.append(float(s))
        except ValueError:
            pass
    return cleaned


def compute_average(scores: list[float]) -> float:
    """Return the average of a list of numbers."""
    return sum(scores) / len(scores)


raw_scores = ["80", "92", "not_a_number", "75"]
print(compute_average(clean_scores(raw_scores)))
````

The second version has a small extra benefit beyond readability: `clean_scores()` can now be reused anywhere you need cleaned data but not an average, and `compute_average()` can be reused on any list of numbers, not just scores.

A quick way to check yourself: try to summarize what a function does in one short sentence, without using "and." If you can't avoid the "and" (e.g. "it cleans the data **and** computes the average"), that's usually a sign it should be split into two functions.

## 6. Reading errors and debugging

Errors are a normal, constant part of writing code! The goal isn't to avoid them, it's to read them efficiently. Let's look at a few common ones:

````python
numbers = [1, 2, 3]
print(numbers[5])
````

Python's error messages (called **tracebacks**) are more helpful than they look at first. Read them **from the bottom up**:
1. The last line gives the **error type** and a short message: here, `IndexError: list index out of range`.
2. The lines above show **where** in your code it happened.
A few error types you'll see constantly:

````python
# NameError: using a variable that doesn't exist
print(undefined_variable)
````

````python
# TypeError: mixing incompatible types
print("age: " + 25)
````

````python
# KeyError: accessing a dictionary key that doesn't exist
teacher = {"name": "William"}
print(teacher["age"])
````

When something doesn't work as expected but doesn't raise an error either, the simplest debugging tool is still often the best one: sprinkle `print()` statements to inspect what your variables actually contain at each step.

````python
def apply_discount(price, discount):
    print(f"price={price}, discount={discount}")   # debug print
    new_price = price * (1 - discount)
    print(f"new_price={new_price}")                  # debug print
    return new_price

apply_discount(100, 0.2)
````

This is deliberately simple, but it's genuinely how most debugging happens day to day: form a hypothesis about what might be wrong, print the relevant variable, confirm or rule it out, repeat. More sophisticated debugging tools exist (like your editor's built-in debugger), but this habit will serve you well before you need anything fancier.

## 7. Using imports correctly

Do you remember imports? It helped you re-use functions (but also constants and classes) from another module rather than having to re-define them.

Did you know that you can import all the content from a module very easily? You just have to use the following line (here, the example is given with the `math` module but the same syntax would work woth any module):

````python
from math import *
````

Now you can use functions from `math` without even having to refer to it (use `sqrt(...)` instead of `math.sqrt(...)`).

````python
print(sqrt(9))
print(cos(pi/3))
````

Now that you know that this feature exists, I'm going to ask you to **forget about it completely**! Using `from module import *` is *never* a good idea for 2 reasons:

- First, using the `from module import ...` syntax imports the content of `module` into the common namespace, without having to type the prefix `module.` to be able to use something from it. This may seem more easy to write but it is also much harder to read! Having to explicitly declare from which module a function is imported often improves maintainability of your code.
- Second, some modules may define the same names, which means that all the definitions of a given name will be erased except that of the most recent import. For example:

````python
from numpy import *
from math import *

angles = [pi, pi/2, pi/3, pi/4]
cos(angles)
````

Here, we wanted to use `pi` from `math` and `cos` from `numpy`, which can be applied to iterables. But since `math` also defines `cos`, its implementations shadows that of `numpy`, leading to this error.
