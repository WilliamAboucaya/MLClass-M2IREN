## 1. Defining a function

A function is defined with `def`, a name, parentheses for its inputs (**parameters**), and a colon. The code inside is indented, just like with `if` and loops.

```python
def greet():
    print("Hello!")

greet()
greet()   # you can call it as many times as you like
```

Functions become useful once they take **inputs**:

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Alex")
greet("Sam")
```

And produce **outputs** with `return`:

```python
def square(x):
    return x ** 2

result = square(4)
print(result)
```

**Important**: `return` immediately exits the function and sends back a value. It's different from `print()`, which just displays something without giving you a value you can reuse. Compare:

```python
def add_v1(a, b):
    print(a + b)   # displays the result, but returns nothing usable

def add_v2(a, b):
    return a + b   # returns the result, so it can be stored/reused

x = add_v1(2, 3)   # x will be None!
y = add_v2(2, 3)   # y will be 5

print(x)
print(y)
```

## 2. Default and keyword arguments

Parameters can have **default values**, making them optional when calling the function:

```python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("William")                    # uses the default greeting
greet("William", "Welcome back")    # overrides the default
```

You can also call arguments by name (**keyword arguments**), in any order:

```python
def describe_person(name, age, city="unknown"):
    print(f"{name}, {age} years old, lives in {city}")

describe_person("William", 29, city="Paris")
describe_person(age=30, name="Someone")   # order doesn't matter with keywords
```

A function can also return **multiple values at once**, as a tuple:

```python
def min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_max([4, 8, 1, 9, 3])
print(lowest, highest)
```

## 3. Scope: local vs. global variables

A variable created **inside** a function only exists inside that function, this is called its **local scope**. It doesn't leak out, and it doesn't clash with a variable of the same name outside.

```python
def compute():
    result = 1234   # local to compute()
    return result

print(compute())
print(result)   # this will raise an error: result doesn't exist here
```

Conversely, a function *can* read a variable defined outside it (**global scope**), but this is generally considered bad practice: it makes functions harder to reuse and reason about. Prefer passing values in as parameters and getting results back with `return`.

```python
# Prefer this:
def add_tax(price, tax_rate):
    return price * (1 + tax_rate)

print(add_tax(100, 0.2))

# Over relying on a variable from outside the function:
tax_rate = 0.2

def add_tax_risky(price):
    return price * (1 + tax_rate)   # works, but fragile: depends on external state

print(add_tax_risky(100))
```

## 4. The `math` module

`math` provides mathematical functions and constants beyond the basic operators you already know. To use it, you **import** it first.

```python
import math

print(math.sqrt(16))     # square root
print(math.pow(2, 10))   # power (as a float)
print(math.floor(4.7))   # round down
print(math.ceil(4.2))    # round up
print(math.pi)           # the constant π
```

Once imported, you access everything inside it with `math.something`. This is called **namespacing**, and it avoids naming clashes between different libraries.

```python
radius = 5
area = math.pi * radius ** 2
print(round(area, 2))   # round() is built-in, no import needed
```

## 5. The `datetime` module

`datetime` lets you work with dates and times: essential once you start handling real-world, time-stamped data.

```python
from datetime import datetime

now = datetime.now()
print(now)
print(type(now))
```

You can create a specific date, and access its components:

```python
birthday = datetime(1998, 6, 15)   # year, month, day
print(birthday)
print(birthday.year)
print(birthday.month)
print(birthday.day)
```

Subtracting two dates gives you a `timedelta`, representing a duration:

```python
start = datetime(2024, 1, 1)
end = datetime(2024, 3, 15)

difference = end - start
print(difference)
print(difference.days)
```

You can also format a date as a readable string with `.strftime()`:

```python
now = datetime.now()
print(now.strftime("%Y-%m-%d"))     # e.g. 2026-07-10
print(now.strftime("%d/%m/%Y"))     # e.g. 10/07/2026
print(now.strftime("%A, %B %d"))    # e.g. Friday, July 10
```

## 6. The `random` module

`random` lets you generate random numbers and make random selections: useful for simulations, sampling, shuffling data, and testing.

```python
import random

print(random.random())          # random float between 0.0 and 1.0
print(random.randint(1, 6))     # random integer between 1 and 6 (inclusive)
print(random.uniform(1, 10))    # random float between 1 and 10
```

You can also pick from, or shuffle, a collection:

```python
fruits = ["apple", "banana", "cherry", "date"]

print(random.choice(fruits))          # one random element
print(random.sample(fruits, 2))       # 2 unique random elements
random.shuffle(fruits)                # shuffles the list in place
print(fruits)
```

Since results are random, re-running these cells will give different outputs each time. If you ever need **reproducible** randomness (e.g. for consistent test results), you can fix the "seed":

```python
random.seed(1234)
print(random.randint(1, 100))   # will always print the same number when seed=1234
```
